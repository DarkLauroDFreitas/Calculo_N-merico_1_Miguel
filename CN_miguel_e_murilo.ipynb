```python
import math
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define uma função para truncar um número em determinada quantidade de casas.
def truncar(numero, casas):
    # Calcula o fator que desloca a vírgula decimal.
    fator = 10 ** casas

    # Trunca o número deslocado e devolve a vírgula à posição original.
    return math.trunc(numero * fator) / fator


# Define uma função para calcular o erro absoluto.
def erro_absoluto(valor_real, valor_aprox):
    # Diferença (em módulo) entre o valor real e o valor aproximado.
    return abs(valor_real - valor_aprox)


# Define uma função para calcular o erro relativo.
def erro_relativo(valor_real, erro_abs):
    # Erro absoluto dividido pelo valor real.
    return erro_abs / valor_real


# Define uma função para calcular o erro percentual.
def erro_percentual(erro_rel):
    # Erro relativo convertido pra porcentagem.
    return 100 * erro_rel
```

## Parte A - truncamento e arredondamento

```python
x1 = 3.14159265
x2 = 98.76543
x3 = 0.00498765

for i in range(5):
  x1_trunc = truncar(x1, i)
  x2_trunc = truncar(x2, i)
  x3_trunc = truncar(x3, i)

  x1_arred = round(x1, i)
  x2_arred = round(x2, i)
  x3_arred = round(x3, i)

  print(f"Truncando em {i} casas - {x1_trunc}, {x2_trunc} {x3_trunc}")

  print(f"Arredondado em {i} casas - {x1_arred}, {x2_arred}, {x3_arred}")

  # Calcular Ea, Er e E%;

  #fazendo contas de erro absoluto

  EAT_x1 = erro_absoluto(x1, x1_trunc)
  EAT_x2 = erro_absoluto(x2, x2_trunc)
  EAT_x3 = erro_absoluto(x3, x3_trunc)

  print(f"Erros absolutos (truncados) -> {EAT_x1} , {EAT_x2}, {EAT_x3}")

  EAA_x1 = erro_absoluto(x1, x1_arred)
  EAA_x2 = erro_absoluto(x2, x2_arred)
  EAA_x3 = erro_absoluto(x3, x3_arred)

  print(f"Erros absolutos (aproximados) -> {EAA_x1} , {EAA_x2} , {EAA_x3}")

 #fazendo conta de erro relativo

  ERT_x1 = erro_relativo(x1, EAT_x1)
  ERT_x2 = erro_relativo(x2, EAT_x2)
  ERT_x3 = erro_relativo(x3, EAT_x3)

  print(f"Erros relativos (truncados) -> {ERT_x1} , {ERT_x2} , {ERT_x3}")

  ERA_x1 = erro_relativo(x1, EAA_x1)
  ERA_x2 = erro_relativo(x2, EAA_x2)
  ERA_x3 = erro_relativo(x3, EAA_x3)

  print(f"Erros relativos (aproximados) -> {ERA_x1} , {ERA_x2} , {ERA_x3}")

 #fazendo conta de erro percentual

  print(f"Erros percentuais (truncados) -> {erro_percentual(ERT_x1)} , {erro_percentual(ERT_x2)} , {erro_percentual(ERT_x3)}")

  print(f"Erros percentuais (aproximados) -> {erro_percentual(ERA_x1)} , {erro_percentual(ERA_x2)} , {erro_percentual(ERA_x3)}")
```

```python
# Graficos da Parte A, item 4: Ea em funcao de n, escala log

ns = list(range(5))

EAT_x1_arr, EAT_x2_arr, EAT_x3_arr = [], [], []
EAA_x1_arr, EAA_x2_arr, EAA_x3_arr = [], [], []

for n in ns:
  EAT_x1_arr.append(erro_absoluto(x1, truncar(x1, n)))
  EAT_x2_arr.append(erro_absoluto(x2, truncar(x2, n)))
  EAT_x3_arr.append(erro_absoluto(x3, truncar(x3, n)))

  EAA_x1_arr.append(erro_absoluto(x1, round(x1, n)))
  EAA_x2_arr.append(erro_absoluto(x2, round(x2, n)))
  EAA_x3_arr.append(erro_absoluto(x3, round(x3, n)))

fig, axs = plt.subplots(1, 3, figsize=(15, 4.5))

dados_plot = [
    ("x1 (3.14159265)", EAT_x1_arr, EAA_x1_arr),
    ("x2 (98.76543)", EAT_x2_arr, EAA_x2_arr),
    ("x3 (0.00498765)", EAT_x3_arr, EAA_x3_arr),
]

for ax, (nome, eat_arr, eaa_arr) in zip(axs, dados_plot):
  # troca 0 por um valor bem pequeno pra nao quebrar a escala log (Ea pode zerar em n=4)
  eat_plot = [v if v > 0 else 1e-16 for v in eat_arr]
  eaa_plot = [v if v > 0 else 1e-16 for v in eaa_arr]

  ax.semilogy(ns, eat_plot, marker="o", label="truncado", color="#a4243b")
  ax.semilogy(ns, eaa_plot, marker="o", label="arredondado", color="#2b6777")
  ax.set_xlabel("n (casas decimais)")
  ax.set_ylabel("Ea (escala log)")
  ax.set_title(nome, fontsize=10)
  ax.set_xticks(ns)
  ax.legend(fontsize=8)
  ax.grid(True, which="both", alpha=0.3)

fig.suptitle("Parte A - Erro absoluto (Ea) em funcao de n, escala logaritmica")
fig.tight_layout()
plt.savefig("parteA_erro_absoluto_log.png", dpi=150)
plt.show()
```

nunca maior. Dá pra ver isso nos prints de cima: quando os dois erram diferente, o truncamento
sempre erra mais; quando o primeiro algarismo cortado é menor que 5, os dois dão exatamente igual.

## Parte C - vazão em um tubo de pequeno diâmetro (Hagen-Poiseuille)

```python
def escoamento(r, L, mi, dp):
  return ((math.pi* (r**4)) * dp) / (8*mi*L)


r = 0.8   # mm
L = 0.2   # m
mi = 3.5 * 10**-3   # Pa.s
dp = 1200   # Pa

r_m = r / 1000  # CUIDADO: r tá em mm e a fórmula pede metro, sem converter Q dá um valor absurdo

Q_referencia = escoamento(r_m, L, mi, dp)

Q_r_arred = escoamento(round(r, 1)/1000, L, mi, dp)
Q_r_trunc = escoamento(truncar(r,1)/1000, L, mi, dp)

Q_mi_arred = escoamento(r_m, L, round(mi, 3), dp)
Q_mi_trunc = escoamento(r_m, L, truncar(mi,3), dp)

Q_dp_arred = escoamento(r_m, L, mi, round(dp, -2))
Q_dp_trunc = escoamento(r_m, L, mi, truncar(dp, -2))

Q_todas_arred = escoamento(round(r,1)/1000, L, round(mi, 3), round(dp, -2))
Q_todas_trunc = escoamento(truncar(r,1)/1000, L, truncar(mi, 3), truncar(dp, -2))

print(f"Valor de Q = {Q_referencia}")
print(f"Q (r arredondado e truncado) = {Q_r_arred} , {Q_r_trunc}")
print(f"Q (mi arredondado e truncado) = {Q_mi_arred} , {Q_mi_trunc}")
print(f"Q (dp arredondada e truncada) = {Q_dp_arred}, {Q_dp_trunc}")
print(f"Q (todas as aprox) = {Q_todas_arred}, {Q_todas_trunc}")


# calculando os erros de cada cenário em relação a Q_referencia (item que tava faltando)

cenarios = {
  "r arred": Q_r_arred, "r trunc": Q_r_trunc,
  "mi arred": Q_mi_arred, "mi trunc": Q_mi_trunc,
  "dp arred": Q_dp_arred, "dp trunc": Q_dp_trunc,
  "todas arred": Q_todas_arred, "todas trunc": Q_todas_trunc,
}

for nome, Q_cenario in cenarios.items():
  Ea = erro_absoluto(Q_referencia, Q_cenario)
  Er = erro_relativo(Q_referencia, Ea)
  Ep = erro_percentual(Er)
  print(f"{nome} -> Ea={Ea:.4e} | Er={Er:.5f} | Ep={Ep:.3f}%")
```

```python
# PARTE C.2 - sensibilidade de Q ao raio (variando de 0.70 a 0.90 mm)

raios = np.arange(0.70, 0.901, 0.01)  # 0.91 dava 1 ponto a mais por imprecisão do float

for r_atual in raios:
  r_atual_m = r_atual / 1000

  Q_ref = escoamento(r_atual_m, L, mi, dp)
  Q_arred = escoamento(round(r_atual, 1)/1000, L, mi, dp)
  Q_trunc = escoamento(truncar(r_atual, 1)/1000, L, mi, dp)

  Ea_arred = erro_absoluto(Q_ref, Q_arred)
  Ea_trunc = erro_absoluto(Q_ref, Q_trunc)

  Er_arred = erro_relativo(Q_ref, Ea_arred)
  Er_trunc = erro_relativo(Q_ref, Ea_trunc)

  print(f"r={r_atual:.2f}mm -> Q_ref={Q_ref:.4e} | Ea_arred={Ea_arred:.3e} Ea_trunc={Ea_trunc:.3e} | Ep_arred={erro_percentual(Er_arred):.3f}% Ep_trunc={erro_percentual(Er_trunc):.3f}%")

```

```python
# Graficos da Parte C.2 - sensibilidade de Q ao raio

Q_ref_C2, Q_arred_C2, Q_trunc_C2 = [], [], []
Ea_arred_C2, Ea_trunc_C2 = [], []
Er_arred_C2, Er_trunc_C2 = [], []

for r_atual in raios:
  r_atual_m = r_atual / 1000

  q_ref = escoamento(r_atual_m, L, mi, dp)
  q_arred = escoamento(round(r_atual, 1)/1000, L, mi, dp)
  q_trunc = escoamento(truncar(r_atual, 1)/1000, L, mi, dp)

  ea_a = erro_absoluto(q_ref, q_arred)
  ea_t = erro_absoluto(q_ref, q_trunc)

  Q_ref_C2.append(q_ref)
  Q_arred_C2.append(q_arred)
  Q_trunc_C2.append(q_trunc)
  Ea_arred_C2.append(ea_a)
  Ea_trunc_C2.append(ea_t)
  Er_arred_C2.append(erro_relativo(q_ref, ea_a))
  Er_trunc_C2.append(erro_relativo(q_ref, ea_t))

Q_ref_C2 = np.array(Q_ref_C2); Q_arred_C2 = np.array(Q_arred_C2); Q_trunc_C2 = np.array(Q_trunc_C2)
Ea_arred_C2 = np.array(Ea_arred_C2); Ea_trunc_C2 = np.array(Ea_trunc_C2)
Er_arred_C2 = np.array(Er_arred_C2); Er_trunc_C2 = np.array(Er_trunc_C2)
Ep_arred_C2 = 100 * Er_arred_C2
Ep_trunc_C2 = 100 * Er_trunc_C2

fig, axs = plt.subplots(2, 2, figsize=(12, 9))

axs[0,0].plot(raios, Q_ref_C2, marker="o", markersize=3, label="Q referencia", color="#023047")
axs[0,0].plot(raios, Q_arred_C2, marker="o", markersize=3, label="Q (r arredondado)", color="#2b6777")
axs[0,0].plot(raios, Q_trunc_C2, marker="o", markersize=3, label="Q (r truncado)", color="#a4243b")
axs[0,0].set_xlabel("raio (mm)"); axs[0,0].set_ylabel("Q (m3/s)")
axs[0,0].set_title("Q em funcao de r"); axs[0,0].legend(fontsize=8)

axs[0,1].plot(raios, Ea_arred_C2, marker="o", markersize=3, label="arredondado", color="#2b6777")
axs[0,1].plot(raios, Ea_trunc_C2, marker="o", markersize=3, label="truncado", color="#a4243b")
axs[0,1].set_xlabel("raio (mm)"); axs[0,1].set_ylabel("Erro absoluto de Q (m3/s)")
axs[0,1].set_title("Erro absoluto de Q em funcao de r"); axs[0,1].legend(fontsize=8)

axs[1,0].plot(raios, Er_arred_C2, marker="o", markersize=3, label="arredondado", color="#2b6777")
axs[1,0].plot(raios, Er_trunc_C2, marker="o", markersize=3, label="truncado", color="#a4243b")
axs[1,0].set_xlabel("raio (mm)"); axs[1,0].set_ylabel("Erro relativo de Q")
axs[1,0].set_title("Erro relativo de Q em funcao de r"); axs[1,0].legend(fontsize=8)

axs[1,1].plot(raios, Ep_arred_C2, marker="o", markersize=3, label="arredondado", color="#2b6777")
axs[1,1].plot(raios, Ep_trunc_C2, marker="o", markersize=3, label="truncado", color="#a4243b")
axs[1,1].set_xlabel("raio (mm)"); axs[1,1].set_ylabel("Erro percentual de Q (%)")
axs[1,1].set_title("Erro percentual de Q em funcao de r"); axs[1,1].legend(fontsize=8)

fig.suptitle("Parte C.2 - Sensibilidade de Q ao raio")
fig.tight_layout()
plt.savefig("parteC2_sensibilidade.png", dpi=150)
plt.show()
```

**Por que amplifica:** Q é proporcional a r^4, então um erro de X% no raio vira mais ou menos
4*X% de erro no Q. É por isso que os erros aqui em cima ficam bem maiores do que só
arredondar/truncar o raio sozinho ia sugerir.


## Parte D - escoamento em um nanocateter

```python
r_nano = 50 * 10**-9   # 50 nm em metro
L_nano = 10 * 10**-6   # 10 micrometro em metro
mi_nano = 3.5 * 10**-3
dp_nano = 5000

Q_nano = escoamento(r_nano, L_nano, mi_nano, dp_nano)

print(f"Q do nanocateter (r=50nm) = {Q_nano}")


# PARTE D.1 - erro geométrico

deltas_r = np.arange(0.1, 5.1, 0.1)  # em nm

for dr in deltas_r:
  dr_m = dr * 10**-9

  r_menos = r_nano - dr_m
  r_mais = r_nano + dr_m

  Q_menos = escoamento(r_menos, L_nano, mi_nano, dp_nano)
  Q_mais = escoamento(r_mais, L_nano, mi_nano, dp_nano)

  Ea_menos = erro_absoluto(Q_nano, Q_menos)
  Ea_mais = erro_absoluto(Q_nano, Q_mais)

  Ea_max = max(Ea_menos, Ea_mais)
  Er_max = erro_relativo(Q_nano, Ea_max)
  Ep_max = erro_percentual(Er_max)

  aprox_linear = 4 * (dr_m / r_nano)   # aproximação dQ/Q ~= 4*dr/r

  print(f"dr={dr:.1f}nm -> Q-={Q_menos:.4e} Q={Q_nano:.4e} Q+={Q_mais:.4e} | Er_max={Er_max:.5f} | aprox_linear={aprox_linear:.5f}")
```

```python
# Graficos da Parte D.1 - erro geometrico no nanocateter

Q_menos_D1, Q_mais_D1 = [], []
Ea_max_D1, Er_max_D1 = [], []

for dr in deltas_r:
  dr_m = dr * 1e-9
  q_menos = escoamento(r_nano - dr_m, L_nano, mi_nano, dp_nano)
  q_mais = escoamento(r_nano + dr_m, L_nano, mi_nano, dp_nano)

  ea_menos = erro_absoluto(Q_nano, q_menos)
  ea_mais = erro_absoluto(Q_nano, q_mais)
  ea_max = max(ea_menos, ea_mais)

  Q_menos_D1.append(q_menos)
  Q_mais_D1.append(q_mais)
  Ea_max_D1.append(ea_max)
  Er_max_D1.append(erro_relativo(Q_nano, ea_max))

Q_menos_D1 = np.array(Q_menos_D1); Q_mais_D1 = np.array(Q_mais_D1)
Ea_max_D1 = np.array(Ea_max_D1); Er_max_D1 = np.array(Er_max_D1)
Ep_max_D1 = 100 * Er_max_D1

fig, axs = plt.subplots(2, 2, figsize=(12, 9))

axs[0,0].plot(deltas_r, Q_menos_D1, label="Q(r-)", color="#a4243b")
axs[0,0].axhline(Q_nano, color="#023047", linestyle="--", label="Q(r)")
axs[0,0].plot(deltas_r, Q_mais_D1, label="Q(r+)", color="#2b6777")
axs[0,0].set_xlabel("dr (nm)"); axs[0,0].set_ylabel("Q (m3/s)")
axs[0,0].set_title("Q(r-), Q(r) e Q(r+)"); axs[0,0].legend(fontsize=8)

axs[0,1].plot(deltas_r, Ea_max_D1, color="#a4243b", marker="o", markersize=3)
axs[0,1].set_xlabel("dr (nm)"); axs[0,1].set_ylabel("Erro absoluto maximo de Q (m3/s)")
axs[0,1].set_title("Erro absoluto maximo de Q")

axs[1,0].plot(deltas_r, Er_max_D1, color="#2b6777", marker="o", markersize=3)
axs[1,0].set_xlabel("dr (nm)"); axs[1,0].set_ylabel("Erro relativo maximo de Q")
axs[1,0].set_title("Erro relativo maximo de Q")

axs[1,1].plot(deltas_r, Ep_max_D1, color="#d1a054", marker="o", markersize=3)
axs[1,1].set_xlabel("dr (nm)"); axs[1,1].set_ylabel("Erro percentual maximo de Q (%)")
axs[1,1].set_title("Erro percentual maximo de Q")

fig.suptitle("Parte D.1 - Erro geometrico no nanocateter")
fig.tight_layout()
plt.savefig("parteD1_erro_geometrico.png", dpi=150)
plt.show()
```

**Comparando com a aproximação linear:** pra dr pequeno (perto de 0.1nm) a aproximação
4*dr/r fica bem próxima do erro relativo real. Conforme dr cresce até 5nm o valor real vai
ficando maior que a aproximação linear, porque Q~r^4 não é linear e a aproximação só pega o
termo de primeira ordem.


```python
# PARTE D.2 - precisão computacional (float32 x float64)

r32 = np.float32(r_nano)
r64 = np.float64(r_nano)

r4_32 = r32 ** np.float32(4)
r4_64 = r64 ** np.float64(4)

print(f"r^4 float32 = {str(r4_32)}")  # str() e nao f-string direto, senao o numpy mostra
print(f"r^4 float64 = {r4_64}")        # digito de mais e some com a diferenca de precisao

Q32 = escoamento(r32, np.float32(L_nano), np.float32(mi_nano), np.float32(dp_nano))
Q64 = escoamento(r64, np.float64(L_nano), np.float64(mi_nano), np.float64(dp_nano))

print(f"Q float32 = {str(Q32)}")
print(f"Q float64 = {Q64}")

dif_abs = abs(float(Q64) - float(Q32))
dif_rel = dif_abs / abs(float(Q64))

print(f"diferença absoluta = {dif_abs}")
print(f"diferença relativa = {dif_rel}")

print(f"eps float32 = {np.finfo(np.float32).eps}")
print(f"eps float64 = {np.finfo(np.float64).eps}")


# comparando as duas formas de calcular a variação (delta1 direto x delta2 expandido)

for dr_teste in [5, 1, 0.1, 1e-3, 1e-6, 1e-9, 1e-12]:
  dr_teste_m = dr_teste * 10**-9

  Q_r = escoamento(r_nano, L_nano, mi_nano, dp_nano)
  Q_r_dr = escoamento(r_nano + dr_teste_m, L_nano, mi_nano, dp_nano)

  delta1 = (Q_r_dr - Q_r) / Q_r
  delta2 = (1 + dr_teste_m / r_nano)**4 - 1

  print(f"dr={dr_teste_m:.2e} -> delta1={delta1:.8e} | delta2={delta2:.8e} | diferença={abs(delta1-delta2):.2e}")
```

**Sobre o cancelamento:** quando dr fica bem pequeno, Q(r+dr) e Q(r) ficam quase iguais em
ponto flutuante, e a subtração usada no delta1 perde precisão (cancelamento catastrófico). O
delta2 não subtrai dois números quase iguais, então se mantém estável mesmo com dr bem pequeno -
dá pra ver a diferença entre os dois crescendo conforme dr diminui nos prints de cima.

**D.3:** Quando o tubo fica muito menor (de mm pra nm), a razão entre área de superfície e
volume cresce bastante, e os efeitos de parede/viscosidade (que num tubo grande são
desprezíveis) passam a dominar o escoamento. Isso quer dizer que o modelo de Hagen-Poiseuille
(que assume fluido contínuo) começa a perder validade física nessa escala, mesmo que a conta
continue saindo com várias casas decimais "certas". Ou seja: precisão numérica não é a mesma
coisa que o modelo estar certo fisicamente.

## Conclusão

Em escalas bem diferentes, a atividade mostrou o mesmo padrão se repetindo: arredondar
sempre erra igual ou menos que truncar, seja num número solto (Parte A) ou na média de
medidas reais (Parte B). O problema é que, quando esse erro entra numa fórmula onde a
grandeza aparece elevada a uma potência alta - como o raio elevado a 4 na vazão de
Hagen-Poiseuille -, ele deixa de ser um detalhe pequeno: um erro de poucos centésimos de
milímetro no raio já vira dezenas de por cento de erro na vazão (Parte C), e o mesmo efeito
aparece de novo, ainda mais evidente, na escala do nanocateter (Parte D). Além disso, o
próprio computador introduz erro por conta própria - a diferença entre calcular em float32
e float64 fica na mesma ordem de grandeza do "epsilon" de cada tipo, e duas contas
matematicamente equivalentes (delta1 e delta2) podem se comportar de forma bem diferente
quando uma delas subtrai dois números quase iguais (cancelamento catastrófico). No fim, o
maior aprendizado é que "preciso" e "correto" não são a mesma coisa: o modelo de
Hagen-Poiseuille pode devolver um Q com várias casas decimais certas e, ainda assim, fazer
cada vez menos sentido físico conforme o tubo encolhe pra escala do nanocateter, onde
efeitos de parede e viscosidade que eram desprezíveis passam a dominar o escoamento.
