```python
import numpy as np
np.set_printoptions(precision=6, suppress=True)
```

```python
A = np.array([[2., 1.,  -1.],
              [-3., -1., 2.],
              [-2., 1., 2.]])
b = np.array([8., -11., -3.])
```

# Exercicio 3

```python
#Cria a matriz aumentada
M = np.column_stack([A, b])

print(M)
```

```python
#Limpa a 1 coluna

m21 = M[1, 0] / M[0, 0]
m31 = M[2, 0] / M[0, 0]

M[1, :] -= m21 * M[0, :]
M[2, :] -= m31 * M[0, :]

#Limpa a 2 coluna

m32 = M[2, 1] / M[1, 1]

M[2, :] -= m32 * M[1, :]

print(M)
```

```python
U = M[:, : - 1]
c = M[:, -1]
x = np.zeros(3)
for i in range (2, -1, -1):
  x[i] = (c[i] - U[i, i + 1:] @ x[i + 1:]) / U[i, i]

print(x)
print(A @ x - b)
```
